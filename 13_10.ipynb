```python
#ex 9
matriz = [
    [12, 5, 8],
    [20, 3, 15],
    [7, 25, 10]
]

maior = matriz[0][0]

for linha in matriz:
    for numero in linha:
        if numero > maior:
            maior = numero

print("O maior número é:", maior)
```

```python
#ex 10
notas = [
    [7, 8, 9],
    [6, 5, 8],
    [10, 9, 7]
]

# a) Exibir todas as notas do primeiro aluno
print("Notas do primeiro aluno:", notas[0])

# b) Exibir somente a segunda nota do segundo aluno
print("Segunda nota do segundo aluno:", notas[1][1])

# c) Utilizar um for para exibir a terceira nota de cada aluno
print("Terceira nota de cada aluno:")
for aluno in notas:
    print(aluno[2])

# d) Alterar a primeira nota do terceiro aluno para 8
notas[2][0] = 8

print("Matriz atualizada:")
for aluno in notas:
    print(aluno)
```
