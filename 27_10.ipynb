```python
#1
cor = input("Digite uma cor: ")

match cor:
    case "azul":
        print("Cor azul")
    case _:
        print("Outra cor")
```

```python
#2
numero = int(input("Digite um número (1, 2 ou 3): "))

match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
    case _:
        print("Número inválido")
```

```python
#3
opcao = int(input("Escolha uma bebida (1, 2 ou 3): "))

match opcao:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")
```

```python
#4
dia = input("Digite um dia da semana: ").lower()

match dia:
    case "sábado" | "sabado" | "domingo":
        print("Final de semana")
    case _:
        print("Dia de semana")
```

```python
#5
nota = input("Digite uma nota (A, B ou C): ").upper()

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```
