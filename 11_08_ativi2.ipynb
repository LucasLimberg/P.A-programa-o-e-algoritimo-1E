```python
#Exercício 3 – Maior de dois números

a = int(input("Digite o primeiro número: "))
b = int(input("Digite o segundo número: "))

if a > b:
  print(f"O número {a} é maior que {b}! ")

elif b > a:
  print(f"O número {b} é maior que {a}! ")

elif a == b:
  print(f"Os dois números são iguais!")
```

```python
#Exercício 4 – Situação do aluno

n = float(input("Digite a nota do aluno: "))

if n >= 7:
  print("Aprovado!")

elif n >= 5 and n < 7:
  print("Recuperação!")

elif n < 5:
  print("Reprovado.")
```

```python
#Exercício 5 – Desconto na compra

va = float(input("Digite o valor da compra: "))

if va >= 100:
  print(f"O valor final da compra é de {va - (va * 0.1)} reais")

else:
  print(f"O valor final da compra é de {va} reais")
```

```python
#Exercício 6 – Positivo, Negativo ou Zero

a = int(input("Digite um número: "))

if a == 0:
  print(f"O número: {a} é igual a: ZERO!")

elif a > 0:
  print(f"O número: {a} é: POSITIVO!")

else:
  print(f"O número {a} é: NEGATIVO!")
```

```python
#Exercício 7 – Classificação de idade

a = int(input("Digite sua idade para verificar a sua faixa etária: "))

if a < 12:
  print("Faixa etária: Criança!")

elif a >= 12 and a <= 17:
   print("Faixa etária: Adolescente!")

elif a >= 18 and a <= 59:
   print("Faixa etária: Adulto!")

elif a >= 60:
   print("Faixa etária: Idoso!")
```

```python
#Exercício 8 – Média de duas notas

a = float(input("Digite a primeira nota: "))
b = float(input("Digite a segunda nota: "))

média = (a + b) /2

print(f"A média do aluno é: {média}")
```

```python
#Exercício 9 – Login simples

usu = str(input("Digite o usuário: "))
senha = int(input("Digite a senha: "))

if usu == "admin" and senha == 1234:
  print("Login realizado com sucesso")

else:
   print("Usuário ou senha incorretos")
```

```python
#Exercício 10 – Calculadora de IMC

peso = float(input("Digite o seu peso: "))
altura = float(input("Digite a sua altura: "))

IMC = peso / (altura * altura)

if IMC < 18.5:
  print("Abaixo do peso!")

elif IMC >= 18.5 and IMC <= 24.9:
  print("Peso normal!")

elif IMC >= 25 and IMC <= 29.9:
  print("Sobrepeso!")

elif IMC >= 30:
  print("Obesidade!")
```
